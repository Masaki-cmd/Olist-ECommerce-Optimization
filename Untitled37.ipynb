## データの準備

まず、アップロードされた`archive.zip`ファイルを解凍し、中身を確認します。これにより、どのようなデータが利用可能か把握できます。

````python
import zipfile
import os

# ZIPファイルのパス
zip_file_path = '/content/archive.zip'

# 解凍先のディレクトリ
extract_dir = '/content/extracted_data'

# ディレクトリが存在しない場合は作成
os.makedirs(extract_dir, exist_ok=True)

# ZIPファイルを解凍
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

print(f"'{zip_file_path}' を '{extract_dir}' に解凍しました。")

# 解凍されたファイルの一覧を表示
print("解凍されたファイル:")
for root, dirs, files in os.walk(extract_dir):
    for file in files:
        print(os.path.join(root, file))
````

## データの読み込みと初期確認

解凍されたCSVファイルをpandasのDataFrameとして読み込み、各データの構造と内容を確認します。まずは主要なデータセットから読み込み、全体の把握に努めます。

````python
import pandas as pd

# 解凍されたファイルのパス
extracted_files = [
    '/content/extracted_data/olist_order_reviews_dataset.csv',
    '/content/extracted_data/olist_products_dataset.csv',
    '/content/extracted_data/olist_geolocation_dataset.csv',
    '/content/extracted_data/olist_sellers_dataset.csv',
    '/content/extracted_data/olist_order_items_dataset.csv',
    '/content/extracted_data/product_category_name_translation.csv',
    '/content/extracted_data/olist_orders_dataset.csv',
    '/content/extracted_data/olist_customers_dataset.csv',
    '/content/extracted_data/olist_order_payments_dataset.csv'
]

# ファイル名をキーとしてDataFrameを格納する辞書
dataframes = {}

# 各CSVファイルを読み込み、DataFrameの情報を表示
for file_path in extracted_files:
    df_name = os.path.basename(file_path).replace('.csv', '')
    print(f"\n--- {df_name} --- ")
    try:
        df = pd.read_csv(file_path)
        dataframes[df_name] = df
        print("Shape:", df.shape)
        print("Columns:", df.columns.tolist())
        print("First 5 rows:\n", df.head())
        print("Missing values:\n", df.isnull().sum())
    except Exception as e:
        print(f"Error reading {df_name}: {e}")
````

## データクリーニング：日付/時刻カラムの変換

データセット全体で日付や時刻を表すカラムが多く存在します。これらを正確に分析するためには、`object`型から`datetime`型に変換する必要があります。これにより、時間の経過に伴うトレンド分析や期間計算が可能になります。

````python
# olist_orders_dataset の日付カラムをdatetime型に変換
print("Converting 'olist_orders_dataset' date columns...")
dataframes['olist_orders_dataset']['order_purchase_timestamp'] = pd.to_datetime(dataframes['olist_orders_dataset']['order_purchase_timestamp'])
dataframes['olist_orders_dataset']['order_approved_at'] = pd.to_datetime(dataframes['olist_orders_dataset']['order_approved_at'])
dataframes['olist_orders_dataset']['order_delivered_carrier_date'] = pd.to_datetime(dataframes['olist_orders_dataset']['order_delivered_carrier_date'])
dataframes['olist_orders_dataset']['order_delivered_customer_date'] = pd.to_datetime(dataframes['olist_orders_dataset']['order_delivered_customer_date'])
dataframes['olist_orders_dataset']['order_estimated_delivery_date'] = pd.to_datetime(dataframes['olist_orders_dataset']['order_estimated_delivery_date'])
print("'olist_orders_dataset' date columns converted. Example:\n", dataframes['olist_orders_dataset'][['order_purchase_timestamp', 'order_delivered_customer_date']].head())

# olist_order_items_dataset の日付カラムをdatetime型に変換
print("\nConverting 'olist_order_items_dataset' date columns...")
dataframes['olist_order_items_dataset']['shipping_limit_date'] = pd.to_datetime(dataframes['olist_order_items_dataset']['shipping_limit_date'])
print("'olist_order_items_dataset' date columns converted. Example:\n", dataframes['olist_order_items_dataset'][['shipping_limit_date']].head())

# olist_order_reviews_dataset の日付カラムをdatetime型に変換
print("\nConverting 'olist_order_reviews_dataset' date columns...")
dataframes['olist_order_reviews_dataset']['review_creation_date'] = pd.to_datetime(dataframes['olist_order_reviews_dataset']['review_creation_date'])
dataframes['olist_order_reviews_dataset']['review_answer_timestamp'] = pd.to_datetime(dataframes['olist_order_reviews_dataset']['review_answer_timestamp'])
print("'olist_order_reviews_dataset' date columns converted. Example:\n", dataframes['olist_order_reviews_dataset'][['review_creation_date', 'review_answer_timestamp']].head())

print("\nAll specified date columns have been converted to datetime objects.")
````

## データクリーニング：商品カテゴリの翻訳と欠損値処理

`olist_products_dataset`には、商品カテゴリ名がポルトガル語で記述されており、一部欠損値も存在します。分析を容易にするため、`product_category_name_translation`データセットを用いて英語に翻訳し、欠損値に対しては適切な処理を行います。

### 翻訳処理
まずは、商品カテゴリ名を英語に翻訳します。これにより、カテゴリごとの売上や利益を分析する際に、より理解しやすい形式になります。

````python
# 商品カテゴリの英語翻訳をマージ
print("Merging product category name translation...")
dataframes['olist_products_dataset'] = pd.merge(
    dataframes['olist_products_dataset'],
    dataframes['product_category_name_translation'],
    on='product_category_name',
    how='left'
)
print("Product categories translated. Example:\n", dataframes['olist_products_dataset'][['product_category_name', 'product_category_name_english']].head())

# 翻訳後の欠損値を確認
missing_translated_categories = dataframes['olist_products_dataset']['product_category_name_english'].isnull().sum()
print(f"\nMissing values in 'product_category_name_english' after merge: {missing_translated_categories}")

# 欠損値の処理：'product_category_name' (ポルトガル語) も欠損している行を確認
# これらの行は、翻訳元となるカテゴリ名自体が存在しないケースです。
# 一旦、不明なカテゴリとして扱います。
# product_category_name_english が NaN の場合、元の product_category_name も NaN の可能性があります。
# 実際には、元の product_category_name が NaN でなければ、翻訳データにないカテゴリということになります。
# 今回のデータでは、product_category_name 自体も610個欠損しているため、これらを 'unknown' として埋めます。

dataframes['olist_products_dataset']['product_category_name_english'].fillna('unknown', inplace=True)
dataframes['olist_products_dataset']['product_category_name'].fillna('unknown', inplace=True)

print(f"Missing values in 'product_category_name_english' after filling: {dataframes['olist_products_dataset']['product_category_name_english'].isnull().sum()}")
print(f"Missing values in 'product_category_name' after filling: {dataframes['olist_products_dataset']['product_category_name'].isnull().sum()}")
print("Filled missing product categories with 'unknown'.")
````

## データ結合：主要データセットの統合

ここまでのデータクリーニングを経て、分析の基盤となる主要なデータセットを結合します。これにより、注文、商品、顧客、販売者、支払いといった様々な情報が紐付けられ、収益に関する課題発見のための多角的な分析が可能になります。

具体的には、以下の順序で結合を進めます。

1.  `olist_orders_dataset` と `olist_order_items_dataset` を結合 (注文と注文品目)
2.  上記の結合結果と `olist_products_dataset` を結合 (商品情報)
3.  さらに `olist_order_payments_dataset` を結合 (支払い情報)
4.  `olist_customers_dataset` を結合 (顧客情報)
5.  `olist_sellers_dataset` を結合 (販売者情報)

````python
print("Merging olist_orders_dataset and olist_order_items_dataset...")
# ordersとorder_itemsをorder_idで結合
df_main = pd.merge(
    dataframes['olist_orders_dataset'],
    dataframes['olist_order_items_dataset'],
    on='order_id',
    how='left'
)
print(f"Shape after merging orders and order_items: {df_main.shape}")

print("\nMerging with olist_products_dataset...")
# main_dfとproductsをproduct_idで結合
df_main = pd.merge(
    df_main,
    dataframes['olist_products_dataset'],
    on='product_id',
    how='left'
)
print(f"Shape after merging products: {df_main.shape}")

print("\nMerging with olist_order_payments_dataset...")
# main_dfとpaymentsをorder_idで結合
df_main = pd.merge(
    df_main,
    dataframes['olist_order_payments_dataset'],
    on='order_id',
    how='left'
)
print(f"Shape after merging payments: {df_main.shape}")

print("\nMerging with olist_customers_dataset...")
# main_dfとcustomersをcustomer_idで結合
df_main = pd.merge(
    df_main,
    dataframes['olist_customers_dataset'],
    on='customer_id',
    how='left'
)
print(f"Shape after merging customers: {df_main.shape}")

print("\nMerging with olist_sellers_dataset...")
# main_dfとsellersをseller_idで結合
df_main = pd.merge(
    df_main,
    dataframes['olist_sellers_dataset'],
    on='seller_id',
    how='left'
)
print(f"Shape after merging sellers: {df_main.shape}")

# 新たに olist_order_reviews_dataset をマージ
print("\nMerging with olist_order_reviews_dataset...")
df_main = pd.merge(
    df_main,
    dataframes['olist_order_reviews_dataset'],
    on='order_id',
    how='left'
)
print(f"Shape after merging order_reviews: {df_main.shape}")

print("\nInitial rows of the combined DataFrame (df_main):\n", df_main.head())
print("\nMissing values in the combined DataFrame (top 20):\n", df_main.isnull().sum().sort_values(ascending=False).head(20))
````

## データクリーニング：結合後の追加処理と特徴量エンジニアリング

主要なデータセットを結合したことで、広範囲にわたる情報が一元化されました。しかし、この結合されたデータには、引き続き欠損値の処理や、分析に役立つ新たな特徴量の作成が必要です。特に、収益分析のKGIに直結する**利益（Profit）**の概念を導入し、配送に関連する時間的特徴量も作成します。

### 1. 欠損値の再確認と処理
結合によって新たな欠損値が発生している可能性があります。特に数値データについては、平均値や中央値での補完、またはビジネスロジックに基づいた処理を検討します。

### 2. 利益関連の特徴量エンジニアリング
KGIである「収益の最大化」に焦点を当てるため、`profit`（利益）を計算するカラムを作成します。このデータセットには直接的な利益情報がないため、`price`（商品価格）と`freight_value`（送料）から粗利益を近似的に算出します。ただし、仕入れコストや運用コストは含まれないため、「粗利益」として扱います。

### 3. 時間関連の特徴量エンジニアリング
注文から配送までの時間や、予定配送日と実績配送日の差異など、時間に関する特徴量は顧客満足度や物流効率の分析に重要です。これらを計算し、新たなカラムとして追加します。

````python
print("\n--- Combined DataFrame Cleaning and Feature Engineering ---")

# 欠損値の処理: payment_typeなど、結合によって生じた欠損の確認
# まずは主要な欠損値を再確認
print("Missing values in df_main before specific handling (top 10):\n", df_main.isnull().sum().sort_values(ascending=False).head(10))

# product_weight_g, product_length_cm, product_height_cm, product_width_cm は、一部商品情報がないため欠損が発生します。
# これらの欠損値は、商品の平均的な値で補完します。これはあくまで概算であり、実際の分析ではより詳細な欠損理由の特定が必要です。
# product_category_name_englishとproduct_category_nameは既に'unknown'で埋められているはずです。

product_dimensions_cols = ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm', 'product_name_lenght', 'product_description_lenght', 'product_photos_qty']
for col in product_dimensions_cols:
    if df_main[col].isnull().any():
        median_val = df_main[col].median()
        df_main[col].fillna(median_val, inplace=True)
        print(f"Filled missing values in '{col}' with median: {median_val}")

# order_approved_at, order_delivered_carrier_date, order_delivered_customer_date
# これらは、まだ未承認、未配送の注文で欠損している可能性があります。これらは注文状況を示すため、欠損はそのままにします。
# ただし、これらを使って時間計算を行う場合は、事前に欠損値を含む行を除外するか、特定のロジックで扱う必要があります。

# review_comment_title, review_comment_message はレビューがない場合に欠損しているため、そのまま。
# payment_type, payment_value などは、結合のhow='left'により注文に紐づく支払いがない場合にNaNになりますが、
# 今回のデータセットでは支払いは常に存在するはずなので、ここでの欠損はまれか、マージの問題の可能性。
# 現状では、payment_valueがNaNの場合、その行は支払いが記録されていないとみなし、分析から除外する。
df_main.dropna(subset=['payment_value'], inplace=True)
print(f"Removed rows with missing 'payment_value'. New shape: {df_main.shape}")

print("\n--- Feature Engineering ---")

# 利益 (Profit) の計算: KGIに直結する指標
# ここでは簡略化のため、商品価格から送料を引いたものを「粗利益」とします。
# 実際のビジネスでは、仕入れ原価や販売手数料などが含まれます。
df_main['gross_profit'] = df_main['price'] - df_main['freight_value']
print("Created 'gross_profit' column. Example:\n", df_main[['price', 'freight_value', 'gross_profit']].head())

# 配送関連の時間特徴量
# 注文から承認までの時間 (seconds)
df_main['time_to_approve_sec'] = (df_main['order_approved_at'] - df_main['order_purchase_timestamp']).dt.total_seconds()
# 注文からキャリアへの引き渡しまでの時間 (seconds)
df_main['time_to_carrier_sec'] = (df_main['order_delivered_carrier_date'] - df_main['order_purchase_timestamp']).dt.total_seconds()
# 注文から顧客への配送までの時間 (seconds)
df_main['time_to_delivery_sec'] = (df_main['order_delivered_customer_date'] - df_main['order_purchase_timestamp']).dt.total_seconds()
# 配送遅延 (予定配送日と実績配送日の差) (seconds)
df_main['delivery_delay_sec'] = (df_main['order_delivered_customer_date'] - df_main['order_estimated_delivery_date']).dt.total_seconds()

print("Created time-related features. Example:\n", df_main[['order_purchase_timestamp', 'order_delivered_customer_date', 'order_estimated_delivery_date', 'time_to_delivery_sec', 'delivery_delay_sec']].head())

# 必要に応じて、時間特徴量の欠損値を処理（未配送の注文など）
# 未配送の注文に対しては、上記の時間特徴量はNaNになるため、それらは別途分析対象とするか、NaNを0などで補完する。
# 今回は、配送遅延の分析に焦点を当てるため、配送が完了していない（order_delivered_customer_dateがNaN）行は、一旦、NaNのままにしておくか、あるいは分析から除外することを検討します。
# 現時点では、そのままにしておき、分析時に必要に応じてフィルタリングします。

print("\nFinal missing values in the combined DataFrame (top 20):\n", df_main.isnull().sum().sort_values(ascending=False).head(20))
````

## 課題定義：収益最大化のための主要課題

これまでのデータ探索と概要把握に基づき、収益を最大化するための具体的な課題を定義します。KGIである「収益の最大化」に貢献するためには、以下の点に注目することが重要です。

*   **売上の増加**: より多くの商品を販売し、客単価を上げるための施策。
*   **利益率の向上**: コスト削減や高利益率商品の販売促進。

現時点でのデータから考えられる課題は以下の通りです。

### 課題1: 低い粗利益率の商品カテゴリの存在
`olist_products_dataset`と`olist_order_items_dataset`を結合した`df_main`には、`price`と`freight_value`から算出した`gross_profit`があります。商品カテゴリごとにこの`gross_profit`の分布や合計値を見ることで、**粗利益率の低い商品カテゴリ**が存在する可能性があります。これは、仕入れコストが高い、競合が激しく価格設定が低い、あるいは送料が商品価格に比べて高すぎる、などの問題を示唆します。

**仮説**: 特定の商品カテゴリは、販売数が多いにも関わらず、粗利益が低い傾向にある。その原因は、送料コストの高さ、または競合による低価格戦略にあるのではないか。

**分析方針**: 商品カテゴリごとの総売上、総粗利益、平均粗利益率を算出し、可視化する。特に粗利益率が低いカテゴリを特定する。

### 課題2: 配送遅延による顧客満足度の低下と売上機会損失
`order_delivered_customer_date`と`order_estimated_delivery_date`の差から`delivery_delay_sec`という特徴量を作成しました。配送遅延は顧客満足度を低下させ、リピート購入の機会損失やレビュー評価の悪化に繋がる可能性があります。

**仮説**: 配送遅延が頻繁に発生している注文は、レビュー評価が低く、顧客ロイヤルティの低下を招いている。結果的に、将来的な売上機会を損失しているのではないか。

**分析方針**: 配送遅延の有無とレビュー評価の関係性を分析する。配送遅延が発生している注文の割合や、遅延日数とレビュー評価の相関を可視化する。

### 課題3: 地域間の販売パフォーマンスの不均衡
`customer_state`や`seller_state`の情報があります。地域によって販売数や粗利益に大きな差がある可能性があります。これは、特定の地域でのマーケティング不足、競合の存在、物流インフラの問題などを示唆します。

**仮説**: 特定の地域（州や都市）では、販売が低迷している、または粗利益率が低い。これは、顧客の需要と供給のミスマッチ、あるいは地域ごとの物流コストの差が影響しているのではないか。

**分析方針**: 顧客の地域 (`customer_state`, `customer_city`) ごとに、売上、粗利益、注文数を集計し、可視化する。特にパフォーマンスが低い地域を特定する。

### 次のステップ：課題1「低い粗利益率の商品カテゴリの存在」の分析

まずは、最初の課題である「**低い粗利益率の商品カテゴリの存在**」について、具体的な分析を進めます。商品カテゴリごとの売上と粗利益を比較することで、収益改善の優先順位を明確にします。

### 分析1: 商品カテゴリごとの売上と粗利益

「低い粗利益率の商品カテゴリの存在」という課題に対応するため、各商品カテゴリが全体の売上と粗利益にどの程度貢献しているかを分析します。

**目的**:
1.  売上貢献度が高いにも関わらず、粗利益率が低いカテゴリを特定する。
2.  全体の収益を押し下げている可能性のあるカテゴリを明らかにする。

この分析により、どのカテゴリに焦点を当てて価格戦略やコスト削減策を検討すべきか、具体的な方向性を見出すことができます。

### 配送遅延を多く発生させているセラー向けのパフォーマンス改善プログラム

過去の分析により、一部のセラーが配送遅延を頻繁に発生させており、それが顧客満足度とレビュー評価の低下に直結していることが明らかになりました。これらのセラーに対して、以下の要素を含む具体的なパフォーマンス改善プログラムを提案します。

#### 1. 物流プロセスに関するトレーニングの提供

配送遅延の原因は多岐にわたりますが、セラー側の不適切な物流プロセスが大きな要因となることがあります。以下のトレーニングを通じて、セラーの物流スキルと効率性を向上させます。

*   **配送梱包ガイドライン**: 商品の破損を防ぎ、輸送効率を高めるための適切な梱包方法に関する詳細なガイドラインを提供します。特に、低利益率の商品カテゴリ（例：electronics, furniture_living_room）において、梱包方法の最適化が送料コストに与える影響についても指導します。
*   **発送手続きの効率化**: 注文処理から発送までのリードタイムを短縮するための手続き改善を支援します。これには、注文管理システムの活用、ピッキング・パッキングプロセスの最適化、発送伝票の自動生成などが含まれます。
*   **在庫管理の最適化**: 在庫切れによる遅延を防ぐため、適切な在庫レベルの維持、需要予測に基づいた発注計画、先入れ先出し（FIFO）原則の徹底など、効率的な在庫管理手法を指導します。
*   **配送状況のモニタリングと対応**: 発送後の配送状況をリアルタイムでモニタリングし、遅延の兆候を早期に察知した場合の顧客への proactive な連絡方法や、配送業者との連携方法について教育します。

#### 2. 配送速度と正確性に関するSLA（サービス品質保証）の設定と実施

セラーの配送パフォーマンスを明確な基準で評価し、目標達成を促すためのSLAを導入します。

*   **配送目標値の明確化**: 各商品カテゴリや配送地域に応じた標準配送期間（例：注文から○日以内に出荷、出荷から○日以内に配送完了）を設定し、セラーと共有します。
*   **パフォーマンス指標の定期的なレポート**: セラーごとに、配送遅延率、平均配送日数、遅延時のレビュー評価などのパフォーマンス指標を定期的にレポートし、改善点を特定しやすくします。
*   **違反時のペナルティまたはインセンティブ制度の導入**:
    *   **ペナルティ**: SLA基準を繰り返し満たせないセラーに対しては、一時的な出品制限、手数料率の引き上げ、または契約解除などの措置を検討します。
    *   **インセンティブ**: SLA基準を継続的に上回る優れたパフォーマンスを示すセラーに対しては、手数料率の優遇、検索順位の向上、プロモーション機会の提供などのインセンティブを付与し、モチベーションを高めます。

#### 3. より信頼性の高い代替物流オプションの推奨

セラーが現在の物流パートナーシップで問題を抱えている場合、より効率的で信頼性の高い代替手段を提案します。

*   **パートナーシップを結んでいる他の配送業者への誘導**: パフォーマンスの高い配送パートナーのリストを共有し、切り替えを促進します。特に、特定の地域（SP, RJ州など）での実績が優れている業者を推奨します。
*   **提携しているフルフィルメントサービス（Fulfillment by Platform）の利用促進**: 自社での発送業務に課題を抱えるセラーには、プラットフォームが提供または提携するフルフィルメントサービス（商品の保管、梱包、発送、顧客対応を一括で代行）の利用を推奨します。これにより、セラーは物流業務から解放され、商品の品質向上やマーケティングに注力できるようになります。また、プラットフォーム側は物流品質をより直接的に管理できるようになります。

#### 期待される効果

これらのプログラムを通じて、セラー全体の配送パフォーマンスを底上げし、配送遅延の削減、顧客満足度の向上、そしてレビュー評価の改善に繋がります。結果として、顧客ロイヤルティが向上し、長期的な売上機会損失を防ぎ、プラットフォーム全体の収益最大化に貢献します。

**Reasoning**:
To analyze the impact of product physical characteristics on logistics costs, I will first filter the main dataframe for the specified low-profit categories, calculate product volume, and then derive freight cost metrics per unit weight and volume. Finally, I will visualize these relationships.



````python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 粗利益率が低いと特定されたカテゴリのデータでdf_mainをフィルタリング
low_profit_categories_list = ['electronics', 'furniture_living_room']
df_low_profit_categories = df_main[df_main['product_category_name_english'].isin(low_profit_categories_list)].copy()

print(f"Filtered data for low-profit categories. Shape: {df_low_profit_categories.shape}")

# 2. 各カテゴリ内で、商品ごとの体積を計算
df_low_profit_categories['product_volume_cm3'] = df_low_profit_categories['product_length_cm'] * \
                                                 df_low_profit_categories['product_height_cm'] * \
                                                 df_low_profit_categories['product_width_cm']

# 3. freight_per_kg と freight_per_volume を計算
# product_weight_g をキログラムに変換
df_low_profit_categories['freight_per_kg'] = df_low_profit_categories['freight_value'] / (df_low_profit_categories['product_weight_g'] / 1000)

# freight_per_volume は体積が0の場合にinfになる可能性があるため、0で割るのを避ける
df_low_profit_categories['freight_per_volume'] = df_low_profit_categories.apply(
    lambda row: row['freight_value'] / row['product_volume_cm3'] if row['product_volume_cm3'] > 0 else 0,
    axis=1
)

print("Calculated product volume, freight_per_kg, and freight_per_volume.")

# 4. 各カテゴリについて、freight_per_kgとfreight_per_volumeの統計量を算出
print("\n--- Freight Cost Ratios Statistics by Category ---")
for category in low_profit_categories_list:
    category_df = df_low_profit_categories[df_low_profit_categories['product_category_name_english'] == category]
    print(f"\nCategory: {category}")
    print("Freight per KG (R$/kg):")
    print(category_df['freight_per_kg'].describe())
    print("\nFreight per Volume (R$/cm^3):")
    print(category_df['freight_per_volume'].describe())

# 5. 可視化
# product_weight_g と freight_value の散布図
plt.figure(figsize=(18, 8))
for i, category in enumerate(low_profit_categories_list):
    category_df = df_low_profit_categories[df_low_profit_categories['product_category_name_english'] == category]

    plt.subplot(1, 2, i + 1)
    sns.regplot(x='product_weight_g', y='freight_value', data=category_df,
                scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
    plt.title(f'{category} - Freight Value vs. Product Weight (g)')
    plt.xlabel('Product Weight (g)')
    plt.ylabel('Freight Value (R$)')
    plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# product_volume_cm3 と freight_value の散布図
plt.figure(figsize=(18, 8))
for i, category in enumerate(low_profit_categories_list):
    category_df = df_low_profit_categories[df_low_profit_categories['product_category_name_english'] == category]

    plt.subplot(1, 2, i + 1)
    sns.regplot(x='product_volume_cm3', y='freight_value', data=category_df,
                scatter_kws={'alpha':0.3}, line_kws={'color':'blue'})
    plt.title(f'{category} - Freight Value vs. Product Volume (cm^3)')
    plt.xlabel('Product Volume (cm^3)')
    plt.ylabel('Freight Value (R$)')
    plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

print("Analysis complete. The plots show the relationship between product physical characteristics and freight value.")
````

    *   **地域専門の配送ドライバー育成**: 各地域の地理に精通し、地域の特性（時間帯による交通状況の変化、特定エリアへの進入規制など）を理解した専門ドライバーを育成します。これにより、配送効率と品質を向上させます。
    *   **ラストワンマイル配送の新しい形態**: 自転車や電動キックボードを用いたエコフレンドリーで迅速な配送手段を、都市部の特定のエリアで試験的に導入することも検討します。
